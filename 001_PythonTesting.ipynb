## ☀️ Day 1: Automated Testing with `unittest`

### Why Automated Testing?
When developing software, we often verify our code by manually running print statements or executing functions with dummy data. While this works initially, it does not scale. Automated tests allow us to run verification checks instantly and repeatedly, ensuring that new changes do not break existing functionality (regressing).

### The `unittest` Module
Python comes with a built-in testing framework called `unittest` (inspired by Java's JUnit). It uses an object-oriented approach where tests are grouped inside classes that inherit from `unittest.TestCase`.

### 1. Conceptual Example: Testing a Basic Math Utility

Let's write a simple helper function and a corresponding unit test to see how the framework structures assertions.

````python
import unittest

# The code under test
def add(a, b):
    """Returns the sum of a and b."""
    return a + b

# The Test Case Class
class TestMathOperations(unittest.TestCase):

    def test_add_integers(self):
        """Test that adding two integers works correctly."""
        self.assertEqual(add(1, 2), 3)

    def test_add_floats(self):
        """Test that adding two floats works correctly."""
        self.assertAlmostEqual(add(0.1, 0.2), 0.3, places=7)

# How to run unittest inside a Jupyter/Colab notebook:
if __name__ == '__main__':
    # we pass argv=['first-arg-is-ignored'], exit=False to prevent exiting the notebook kernel
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
````

### 🧠 Key Anatomy of `unittest`
1. **`unittest.TestCase`**: Your test class must inherit from this to access assertions.
2. **`test_` prefix**: Any method intended to run as a test **must** begin with the prefix `test_`. Otherwise, the test runner will ignore it.
3. **Assertions**: Methods like `assertEqual`, `assertTrue`, `assertRaises` are used to check if the expected outcome matches the actual result.

## ☀️ Day 3: Transitioning to `pytest`

Welcome to Day 3! Today, we transition from Python's standard library `unittest` to `pytest`, one of the most popular and powerful testing frameworks in the Python ecosystem.

### Why `pytest`?
While `unittest` is robust, it requires a lot of boilerplate code: you must import `unittest`, create classes that inherit from `unittest.TestCase`, and use specific assertion methods like `self.assertEqual` or `self.assertRaises`.

`pytest` simplifies this significantly:
1. **No Boilerplate Classes**: Tests can be written as simple, top-level functions.
2. **Plain `assert` Statements**: Instead of remembering dozens of custom assertion methods, you use Python's built-in `assert` statement. `pytest` intercepts these and provides rich, detailed failure reports.
3. **Easy Test Discovery**: `pytest` automatically finds and runs any file named `test_*.py` or `*_test.py` and any function prefixed with `test_`.

---

### 🐍 Writing Your First `pytest` Test

Compare how we assert equality between the two frameworks:

**With `unittest`:**
```python
class TestMath(unittest.TestCase):
    def test_add(self):
        self.assertEqual(add(1, 2), 3)
```

**With `pytest`:**
```python
def test_add():
    assert add(1, 2) == 3
```

`pytest` also simplifies exception testing using its own context manager: `pytest.raises`.

```python
import pytest

def test_invalid_input():
    with pytest.raises(TypeError):
        reverse_and_capitalize(123)
```

### 🛠️ Practice Challenge 5: Converting `unittest` to `pytest`

Below, we have a simple function `divide(a, b)` that returns the result of division, or raises a `ValueError` if attempting to divide by zero.

**Your Task:**
Write two simple testing functions using **`pytest` style** (no classes, plain `assert` statements, and `pytest.raises` for exceptions):
1. `test_divide_success()`: Verify that dividing `10` by `2` correctly returns `5.0`.
2. `test_divide_by_zero()`: Verify that dividing `10` by `0` raises a `ValueError` with an appropriate check.

````python
import pytest

# The code under test
def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b

# Implement test_divide_success using plain assert
def test_divide_success():
    assert divide(10, 2) == 5.0

# Implement test_divide_by_zero using pytest.raises
def test_divide_by_zero():
    with pytest.raises(ValueError, match="Cannot divide by zero"):
        divide(10, 0)
````

### 🛠️ `pytest` Fixtures: Injecting Dependencies

In `unittest`, we write `setUp` methods inside a class. In `pytest`, we define **fixtures** as functions with the `@pytest.fixture` decorator. Any test function can use a fixture simply by accepting an argument named after the fixture function!

Here is how we set up a fixture for a database connection:

```python
import pytest

@pytest.fixture
def sample_data():
    return {"username": "alice", "role": "admin"}

def test_user_role(sample_data):
    # pytest automatically executes sample_data() and passes the return value here
    assert sample_data["role"] == "admin"
```

### 🔄 Setup & Teardown with `yield`

If your fixture requires cleanup (like closing a database or deleting a file), you replace the `return` statement with a `yield`. Everything before `yield` is the **setup**, and everything after `yield` is the **teardown**:

```python
@pytest.fixture
def temporary_file():
    # Setup code
    f = open("temp.txt", "w")
    yield f  # Provide the fixture value to the test
    # Teardown code (runs after the test finishes)
    f.close()
```

### 🛠️ Practice Challenge 6: Writing a `pytest` Fixture

Let's rewrite our `ShoppingCart` tests from Day 2 using `pytest` fixtures!

**Your Task:**
1. Define a `pytest` fixture named `empty_cart` that returns a clean instance of `ShoppingCart`.
2. Write two test functions:
   * `test_add_item(empty_cart)`: Use the fixture to add an item to the cart and assert the total price matches.
   * `test_remove_nonexistent_item(empty_cart)`: Use the fixture and verify that removing a nonexistent item raises a `ValueError`.

````python
%%writefile shopping_cart.py
import pytest

# The code under test
class ShoppingCart:
    def __init__(self):
        self.items = {}

    def add_item(self, item_name, price, quantity=1):
        if quantity <= 0:
            raise ValueError("Quantity must be greater than zero")
        if item_name in self.items:
            self.items[item_name]['quantity'] += quantity
        else:
            self.items[item_name] = {'price': price, 'quantity': quantity}

    def remove_item(self, item_name, quantity=1):
        if item_name not in self.items:
            raise ValueError("Item not in cart")
        if quantity <= 0:
            raise ValueError("Quantity must be greater than zero")
        if self.items[item_name]['quantity'] < quantity:
            raise ValueError("Cannot remove more items than present in cart")

        self.items[item_name]['quantity'] -= quantity
        if self.items[item_name]['quantity'] == 0:
            del self.items[item_name]

    def get_total(self):
        return sum(details['price'] * details['quantity'] for details in self.items.values())
````

````python
%%writefile test_shopping_cart.py

from shopping_cart import ShoppingCart

import pytest

@pytest.fixture
def empty_cart():
  return ShoppingCart()

# Implement test_add_item(empty_cart)
def test_add_item(empty_cart):
  empty_cart.add_item("apple", 1.5, 2)
  assert empty_cart.get_total() == 3.0
  assert "apple" in empty_cart.items

# Implement test_remove_nonexistent_item(empty_cart)
def test_remove_nonexistent_item(empty_cart):
  with pytest.raises(ValueError, match="Item not in cart"):
    empty_cart.remove_item("orange", 1)
````

## ☀️ Day 4: Advanced `pytest` (Parametrization & Plugins)

Welcome to Day 4! Now that you understand the core mechanics of `pytest` and how to write fixtures, it's time to supercharge your testing suite using advanced features.

### 🧬 Test Parametrization (`@pytest.mark.parametrize`)

In testing, we often need to run the same test logic multiple times with different inputs and expected outputs.

Instead of writing separate test functions (like `test_add_positives`, `test_add_negatives`, etc.), `pytest` allows us to **parametrize** a single test function using the `@pytest.mark.parametrize` decorator.

Here is how simple it is:

```python
import pytest

@pytest.mark.parametrize("a, b, expected", [
    (1, 2, 3),
    (-1, 1, 0),
    (0, 0, 0),
    (10, 20, 30)
])
def test_add(a, b, expected):
    assert a + b == expected
```

When `pytest` runs, it will dynamically generate and execute **4 separate test cases**, reporting failures and successes individually! This is incredibly useful for testing edge cases in mathematical models, validation pipelines, and text parsing.

### 🛠️ Practice Challenge 7: Parametrizing Your Validation Tests

Let's revisit our `validate_email(email)` utility from Day 1 and rewrite it to use `pytest` with parametrization.

**Your Task:**
1. Define the `validate_email(email)` function.
2. Write a single parametrized test function `test_valid_emails(email, expected_domain)` that tests several valid email inputs and checks that the function returns the correct domain.
3. Write another parametrized test function `test_invalid_emails(email)` that uses `pytest.raises(ValueError)` to assert that invalid email formats correctly raise errors.

````python
%%writefile test_email_parametrize.py
import pytest

# The function under test
def validate_email(email):
    domain_list = ['com', 'org', 'edu']
    if '@' not in email or '.' not in email:
        raise ValueError("Invalid email format")
    domain = email.split('.')[-1]
    if domain not in domain_list:
        raise ValueError("Unsupported domain")
    return domain

# Parametrize this test with valid email scenarios
@pytest.mark.parametrize("email, expected_domain", [
    ("alice@example.com", "com"),
    ("bob@data.org", "org"),
    ("charlie@university.edu", "edu")
])
def test_valid_emails(email, expected_domain):
    assert validate_email(email) == expected_domain

# Parametrize this test with invalid email scenarios
@pytest.mark.parametrize("email", [
    ("no_at_symbol.com"),
    ("username@invaliddomain.net"),
    ("missing_dot@com")
])
def test_invalid_emails(email):
    with pytest.raises(ValueError):
        validate_email(email)
````
